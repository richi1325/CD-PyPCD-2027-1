```python
lista = [-1]
if lista:
    print(len(lista))
else:
    print("Lista vacia")
```

```python
edad = 18
if (edad < 18):
  print("Eres menor de edad")
elif (edad >= 18 and edad <= 30):
  print("Adulto Joven")
else:
  print("Adulto")
```

# Estructuras Reperetitivas

```python
for caracter in "hola":
    print(caracter)
```

```python
for elemento in [1,2,3,4,5]:
    print(elemento)
```

```python
for elemento in [[1,2], [3,4], [5,6]]:
    print(elemento)
```

```python
len([1,2,3,4,5])
```

```python
len([[1,2], [3,4], [5,6]])
```

```python
for i in range(5):
  print(i)
```

```python
nombre = "Ricardo"
for i in range(len(nombre)):
  print(nombre[i])
```

```python
nombre = "Ricardo"
for i in nombre:
  print(i)
```
