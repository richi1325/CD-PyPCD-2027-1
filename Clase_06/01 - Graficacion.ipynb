# Serie de Tiempo

```python
fig, ax = plt.subplots(figsize=(10, 10))
dias = np.arange(1, 31)
ventas_modelo_a = np.random.randint(100, 200, size = 30)
ventas_modelo_b = np.random.randint(80, 220, size = 30)
ax.plot(
    dias,
    ventas_modelo_a,
    color = "#414361",
    linestyle = "-",
    linewidth = 2,
    label = "Modelo A"
)
ax.plot(
    dias,
    ventas_modelo_b,
    color = "#7F2CCB",
    linestyle = "--",
    linewidth = 2,
    label = "Modelo B"
)
ax.set_xticks(dias) # Arreglo con los números del eje X
ax.set_ylim(bottom=0, top=220) # Definición del límite inferior y superior
ax.set_xlabel("Días del mes")
ax.set_ylabel("Unidades vendidas")
ax.set_title("Ventas por día de modelos de negocio")
ax.grid(True, alpha = 0.1) # Alpha es transparencia
ax.legend(loc="upper left")
plt.show()
#plt.savefig("grafica1.png")
```

#  Scatter Plots (Grafica de dispersión)

```python
fig, ax = plt.subplots(figsize=(10, 6))
metros_cuadrados = np.random.normal(170, 40, 200)
precio = 1500 * metros_cuadrados + np.random.normal(0, 20000, 200)
ax.scatter(metros_cuadrados, precio)

x = np.arange(metros_cuadrados.min(), metros_cuadrados.max(), step = 0.5)
y = 1500 * x
ax.plot(
    x,
    y,
    color = "#7F2CCB",
    linestyle = "--",
    linewidth = 2,
)
ax.set_xlabel("Metros Cuadrados")
ax.set_ylabel("Precio (MXN)")
ax.set_title("Precio vivienda por metros cuadrados")
plt.show()
```

# Distribuciones y Outliers

```python
fig, ax = plt.subplots(1, 2, figsize=(10, 6))
salarios = np.random.normal(50000, 8000, 1000)
outliers = [180000, 200000, 3000, 1200]
salarios_completos = np.concat([salarios, outliers])

# Histograma
ax[0].hist(salarios_completos, bins = 40, color = "skyblue", edgecolor = "black")
ax[0].axvline(50000, color = "red", linestyle="--")

# Boxplot
ax[1].boxplot(salarios_completos)
plt.show()
```

```python
fig, ax = plt.subplots(figsize=(10, 6))
ax.violinplot(salarios_completos, vert=False, showextrema=False)
ax.boxplot(
    salarios_completos,
    orientation= "horizontal",
    widths=0.15,
    patch_artist=True,
    boxprops=dict(facecolor="#333333", color="#333333"),
    medianprops=dict(color="white", linewidth=2),
    whiskerprops=dict(color="#333333", linewidth=1.5),
    capprops=dict(color="#333333", linewidth=1.5)
)
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

fig, ax = plt.subplots(figsize=(10, 6))
sns.violinplot(x=salarios_completos, ax=ax, inner="box")

plt.title("Distribución de salarios")
plt.xlabel("Salario")
plt.show()
```
