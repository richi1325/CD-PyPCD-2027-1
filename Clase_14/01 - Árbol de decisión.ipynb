```python
from sklearn.tree import DecisionTreeClassifier, export_text
```

```python
X = [[1 ,50] ,[2 ,60] ,[4 ,85] ,[5 ,75] ,[6 ,85] ,[7 ,95] ,[1 ,90]]
y = [0 ,0 ,1 ,0 ,1 ,1 ,0]
```

```python
arbol = DecisionTreeClassifier(
    criterion = "gini", max_depth = 3 , random_state = 0
)
```

```python
arbol.fit(X, y)
```

```python
print(export_text(arbol, feature_names = ["horas", "asistencia"]))
```

```python
from sklearn.tree import plot_tree
import matplotlib.pyplot as plt
```

```python
plot_tree(
    arbol,
    feature_names =["horas", "asistencia"],
    class_names =["reprobado", "aprobado"],
    filled = True,
    rounded = True
)
plt.show()
```

```python
print(arbol.predict([[5 , 85]]))
```
