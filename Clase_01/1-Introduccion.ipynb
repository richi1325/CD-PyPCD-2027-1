# Cadenas

En esta sección aprenderemos de las funciones para **manipular** _cadenas_

```python
len('hola mundo')
```

```python
saludo = "Bienvenido(a) a la licenciatura en ciencia de datos,"
nombre = "Ricardo"
```

```python
# Concatenación
saludo + " " + nombre
```

```python
# Interpolacion 1
saludo = "Bienvenido(a) {name} a la licenciatura en ciencia de datos, {name}" #perezosa
nombre = "Ricardo"
print(saludo.format(name=nombre))
print(saludo.format(name="Alejandro"))
```

```python
# Interpolacion 2
nombre = "Ricardo"

saludo = f"Bienvenido(a) {nombre} a la licenciatura en ciencia de datos, {nombre}" # f-strings anciosas
print(saludo, end="$")
nombre = "Alejandro"
print(saludo)
```

```python
# Manipulación de cadenas
nombre = "aLEjAndrO ToRReS"
nombre = nombre.upper() #Inmutables
nombre
```

```python
nombre.lower()
```

```python
nombre.capitalize()
```
