```python
# Snake Case
def saludarNombre(nombre: str) -> None:
    """Esta función te da un saludo amigable :)"""
    print(f"Bienvenido(a) a ciencia de datos, {nombre}!")
```

```python
def factorial(n:int) -> int:
    temp:int = 1 # neutro multiplicativo
    for i in range(1, n + 1):
        temp = temp * i
    return temp
```

```python
for i in range(1, 10 + 1):
    resultado_factorial = factorial(i)
    print(f"{i}! = {resultado_factorial}")
```

```python
def suma(*args):
    temp = 0
    for i in args:
        temp = temp + int(i)
    return temp
```

```python
suma(2, 3, 4, "10")
```
