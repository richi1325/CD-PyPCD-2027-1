```python
persona = {
    "nombre": "Ricardo",
    "apellido": "Mendoza",
    "es_profesor": True,
    "edad": 25,
}
```

```python
persona["nombre"] + " " + persona["apellido"]
```

```python
for element in persona.keys():
    print(element)
```

```python
for clave in persona.keys():
    print(persona[clave])
```

```python
for value in persona.values():
    print(value)
```

```python
for llave, valor in persona.items():
    print(llave, valor)
```

```python
persona["carrera"] = "MAC"
```

```python
persona
```

```python
persona["nombre"] = "Jose Ricardo"
```

```python
persona
```

```python
# Acceder a un valor, no seguro
persona["edad2"]
```

```python
# Acceder a un valor, forma segura (no produce errores)
persona.get("edad2", 0) # valor a devolver
```

```python
with open("./data/cd.txt", "r") as f:
    wiki = f.read()
```

```python
wiki = wiki.split()
```

```python
conteo_palabras = dict()
for palabra in wiki:
    conteo_palabras[palabra] = conteo_palabras.get(palabra, 0) + 1
```

```python
with open("./output/resultados.txt", "w") as f:
    f.write(str(conteo_palabras))
```
