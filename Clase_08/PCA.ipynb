```python
df_wines = pd.concat([df_red_wine,df_white_wine], axis = 0).reset_index().drop("index", axis =1)
```

```python
var_numericas = ['fixed acidity', 'volatile acidity', 'citric acid', 'residual sugar',
       'chlorides', 'free sulfur dioxide', 'total sulfur dioxide', 'density',
       'pH', 'sulphates', 'alcohol']
X = df_wines[var_numericas]
Y = df_wines['quality']
```

```python
Y
```

```python
X
```

```python
scaler = StandardScaler()
```

```python
scaler.fit(X)
```

```python
X_transformed = scaler.transform(X)
```

```python
X_transformed
```

```python
pca=PCA(n_components = 2)

```

```python
pca.fit(X_transformed)
```

```python
print(pca.explained_variance_ratio_)
```

```python
X_pca = pca.transform(X_transformed)
```

```python
X_pca
```

```python
df_pca= pd.DataFrame(X_pca, columns=('X_pca1', 'X_pca2'))
```

```python
df_pca
```

```python
df_pca['calidad'] = Y
```

```python
df_pca
```

```python
sns.scatterplot(df_pca, x ='X_pca1', y ='X_pca2', hue = 'calidad', palette="deep")
 
```

# PCA All

```python
pca_all = PCA(n_components = df_wines.shape[1] - 1)
```

```python
pca_all.fit(X_transformed)
```

```python
# Criterio para elegir el tamaño de componentes
suma_acumulada = 0
i = 1
for element in list(pca_all.explained_variance_ratio_):
    suma_acumulada += element
    print(i, suma_acumulada)
    i += 1
# Para este caso fue de 5
```

# PCA 3D

```python
pca3n = PCA(n_components = 3)
```

```python
pca3n.fit(X_transformed)
```

```python
X_pca3n = pca3n.transform(X_transformed)
```

```python
df_pca3n = pd.DataFrame(X_pca3n, columns=('PCA1', 'PCA2', 'PCA3'))
```

```python
df_pca3n['Calidad'] = Y
```

```python
import matplotlib.pyplot as plt

# 1. Tu diccionario de marcadores
marcadores = {
    3: 'o',  # Círculo
    4: 's',  # Cuadrado
    5: '^',  # Triángulo arriba
    6: 'D',  # Diamante
    7: 'p',  # Pentágono
    8: '*',  # Estrella
    9: 'v'   # Triángulo abajo
}

fig = plt.figure(figsize=(8, 6))
ax = fig.add_subplot(projection='3d')

# 2. Agrupar por la columna Calidad e iterar
for calidad_val, grupo in df_pca3n.groupby("Calidad"):
    ax.scatter(
        grupo["PCA1"], 
        grupo["PCA2"], 
        grupo["PCA3"], 
        marker=marcadores.get(calidad_val, 'o'),  # Un solo símbolo por grupo
        label=f'Calidad {calidad_val}'
    )

ax.set_xlabel("PCA1")
ax.set_ylabel("PCA2")
ax.set_zlabel("PCA3")
ax.legend(title="Calidad")

plt.show()
```
