```python
import numpy as np
from sklearn import datasets
from sklearn.neighbors import KNeighborsClassifier
import seaborn as sns
import matplotlib.pyplot as plt
from matplotlib.colors import ListedColormap
```

```python
iris = datasets.load_iris()
```

```python
iris
```

```python
X = iris['data'][:,0:2] # 'sepal length (cm)', 'sepal width (cm)'
y = iris['target']
#y = []
#for el in iris['target']:
#    y.append(str(iris["target_names"][el]))
#y = np.array(y)
```

```python
# 1. Configurar el modelo
k_vecinos = 13 # Hyperparametros
knn = KNeighborsClassifier(n_neighbors = k_vecinos)
```

```python
# 2. Entrenamiento
knn.fit(X, y)
```

```python
# 3. Evaluación del modelo
knn.score(X, y)
```

```python
# 4. Predicción
flor = np.array([
    [5.3, 3.7]
])
str(knn.predict(flor)[0])
```

```python
x_min, x_max = X[:,0].min(), X[:,0].max() + 1
y_min, y_max = X[:,1].min(), X[:,1].max() + 1
xx, yy = np.meshgrid(
    np.arange(x_min, x_max, 0.02),
    np.arange(y_min, y_max, 0.02)                
)
y_predict = knn.predict(np.c_[xx.ravel(), yy.ravel()])
y_predict = y_predict.reshape(xx.shape)
```

```python
xx.shape, yy.shape, y_predict.shape
```

```python
# Visualización
plt.figure(figsize=(10, 6))
cmap_light = ListedColormap(['#FFAAAA', '#AAFFAA', '#AAAAFF'])
cmap_bold = ['red', 'green', 'blue']

# Pintar las áreas de decisión de fondo
plt.contourf(xx, yy, y_predict, cmap=cmap_light, alpha=0.8)

# Dibujar los puntos reales del dataset encima
sns.scatterplot(
    x=X[:, 0], 
    y=X[:, 1], 
    hue=iris.target_names[y],
    palette=cmap_bold, 
    alpha=1.0, 
    edgecolor="black"
)

plt.title(f"Clasificación k-NN en Dataset Iris (k = {k_vecinos})")
plt.xlabel(iris.feature_names[0])
plt.ylabel(iris.feature_names[1])
plt.legend(title="Especies")
plt.show()
```
